```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import cv2 
import os

import imgaug as ia
import imgaug.augmenters as iaa
import albumentations as A
from tqdm import tqdm
```

```python
# Configuratie variablen
INPUT_FOLDER = 'data/dataset'  
OUTPUT_FOLDER = 'dataset_augmented/' 
AUGMENTATIONS_PER_IMAGE = 5 

# Definieer de pipeline voor de soorten augmentatie
transform = A.Compose([
    A.HorizontalFlip(p=0.5),
    A.Rotate(limit=30, p=0.7, border_mode=cv2.BORDER_CONSTANT),
    # Zoom out via Affine (schaal 0.5 tot 0.9)
    A.Affine(scale=(0.5, 0.9), p=0.6, border_mode=cv2.BORDER_CONSTANT),
    # Resolution scaling (maakt het beeld blokkeriger/vager)
    A.Downscale(scale_min=0.25, scale_max=0.5, p=0.4),
])

# Maak de output map aan als deze nog niet bestaat
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# Verwerk de bestanden
image_files = [f for f in os.listdir(INPUT_FOLDER) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

print(f"Start met augmenteren van {len(image_files)} afbeeldingen...")

for filename in tqdm(image_files):
    # Laad origineel
    img_path = os.path.join(INPUT_FOLDER, filename)
    image = cv2.imread(img_path)
    
    if image is None:
        continue
        
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    # Genereer X aantal variaties
    for i in range(AUGMENTATIONS_PER_IMAGE):
        augmented = transform(image=image)['image']
        
        # Converteer terug naar BGR voor OpenCV opslag
        augmented_bgr = cv2.cvtColor(augmented, cv2.COLOR_RGB2BGR)
        
        # Maak een nieuwe bestandsnaam
        new_name = f"aug_{i}_{filename}"
        save_path = os.path.join(OUTPUT_FOLDER, new_name)
        
        cv2.imwrite(save_path, augmented_bgr)

print(f"Klaar! Je dataset is nu uitgebreid naar {len(os.listdir(OUTPUT_FOLDER))} afbeeldingen.")
```
